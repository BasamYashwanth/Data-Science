```python
import seaborn as sns
```

```python
df=sns.load_dataset('titanic')
```

```python
sns.distplot(df['age'],bins=10,color='g',hist=False,kde=True)
```

```python
sns.jointplot(x=df['age'],y=df['fare'],color='yellow',hue=df['sex'])
```

```python
sns.pairplot(df)
```

```python
sns.barplot(x=df['sex'],y=df['age'])
```

```python
sns.countplot(df['age'])
```

```python
sns.boxplot(x=df['age'])
```

```python
sns.violinplot(x=df['age'],y=df['sex'])
```
