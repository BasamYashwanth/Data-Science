```python
import matplotlib.pyplot as plt
```

Create a first chart

```python
x=[1,2,3,4,5]
y=[10,20,30,40,50]
plt.figure(figsize=(4,3))
plt.plot(x,y,color='r',marker='o',linestyle='dashed',linewidth=5,markersize=12) #line chart
plt.title('Line chart')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
```

```python
x=[1,2,3,4,5]
y1=[10,20,30,40,50]
y2=[20,30,40,50,60]
plt.plot(x,y1,label='sales 2025')
plt.plot(x,y2,label='sales 2026')

plt.title("YOY Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
x=['A','B','C','D','E']
y=[10,20,30,40,50]
plt.bar(x,y,color='r')
plt.title('Bar chart')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
```

```python
# used for distribution analysis
# data
import random
data = [random.randint(1, 30) for _ in range(500)]
plt.hist(data, bins=20) # histogram chart
plt.title("Histogram Example")
plt.show()
```

```python
# data
categories = ['A','B','C','D','E']
sales = [10, 20, 55, 35, 45]


plt.pie(sales, labels=categories, autopct = '%1.1f%%', startangle=90)
plt.title("Pie Chart Example")
plt.show()
```

```python
# used to show multiple charts in one figure

# data-1 - bar chart
categories = ['Mon','Tue','Wed','Thu','Fri']
sales = [10, 20, 55, 35, 45]


# data-2 - scatter plot
y1 = [10, 20, 25, 35, 45]
y2 = [20, 30, 35, 45, 55]

plt.figure(figsize=(10,4))
# first plot- bar chart
plt.subplot(1,2,1) # row, column, position
plt.bar(categories, sales)
plt.title("Daily Sales")
plt.xlabel("Week Days")
plt.ylabel("Sales")
# second plot- scatter chart
plt.subplot(1,2,2) # row, column, position
plt.scatter(y1, y2)
plt.title("User Example")
plt.xlabel("User1")
plt.ylabel("User2")

plt.show()
```
