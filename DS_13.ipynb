```python
#Simple Random Sampling
import random as r
population = [1,5,7,3,9,2,0,7,8,2]
sample_size=5
sample = r.sample(population, sample_size)
print(sample)
```

```python
#Stratified Sampling
group_A=[1,6,8,3,0]           #create groups (strata)
group_B=[2,8,4,9,3]
sample_A=r.sample(group_A,2)  #selecct sample from each group
sample_B=r.sample(group_B,2)
sample=sample_A+sample_B      #combine the sample
print(sample)
```

```python
#ystematic Sampling
population =[2,7,1,9,4,7,3,6,0,5,8]
k=2
sample=population[::k]
print(sample)
```

```python
import random

# Population
students = [
    "Alice", "Bob", "Charlie", "David", "Emma",
    "Frank", "George", "Helen", "Ian", "Jack",
    "Kate", "Leo", "Mary", "Nick", "Olivia",
    "Paul", "Queen", "Ryan", "Sarah", "Tom"
]

# Population size
N = len(students)

# Required sample size
n = 5

# Calculate sampling interval
k = N // n

# Choose a random starting position
start = random.randint(0, k - 1)

# Select every k-th student
sample = students[start::k]

print("Sampling interval:", k)
print("Starting position:", start)
print("Selected students:", sample)
```

```python
#Cluster Sampling
import random
cluster_1=[2,6,4,2]
cluster_2=[9,5,3,7]
cluster_3=[3,9,0,6]
cluster=[cluster_1,cluster_2,cluster_3]
select_cluster=random.choice(cluster)
print(select_cluster)
```
