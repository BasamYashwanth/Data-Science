```python
# import numpy for numerical array operations
import numpy as np
# import pandas to hold the sample student performance data
import pandas as pd

# create a small sample dataset: marks of students in 3 subjects
data = pd.DataFrame({
    "Math": [85, 78, 90, 60],
    "Science": [80, 82, 88, 65],
    "English": [75, 70, 92, 58]
}, index=["S1", "S2", "S3", "S4"])

# define a function to compute Euclidean distance between two rows
def euclidean_distance(row1, row2):
    # sqrt of sum of squared differences between corresponding elements
    return np.sqrt(np.sum((row1 - row2) ** 2))

# compute the distance between student S1 and S2
dist_12 = euclidean_distance(data.loc["S1"], data.loc["S2"])
# compute the distance between student S1 and S3
dist_13 = euclidean_distance(data.loc["S1"], data.loc["S3"])

print("Euclidean distance S1-S2:", dist_12)
print("Euclidean distance S1-S3:", dist_13)
```
