```python
import pandas as pd
import seaborn as sns
```

```python
peng = sns.load_dataset("penguins") #iris
```

```python
peng.head()
```

```python
x=peng.drop(columns=["species","island","sex"],axis=1)
```

```python
x.corr().round(2)
```

```python
sns.heatmap(x.corr().round(2),annot=True,cmap="Blues")
```
