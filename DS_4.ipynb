```python
#remove outliers
# Select only numerical columns for quantile calculation
df_numeric = df.select_dtypes(include=np.number)

Q1 = df_numeric.quantile(0.25)
Q3 = df_numeric.quantile(0.75)
IQR = Q3 - Q1
print(IQR)
```

```python
# Remove outliers from numerical columns only
df_no_outliers = df_numeric[~((df_numeric < (Q1 - 1.5 * IQR)) | (df_numeric > (Q3 + 1.5 * IQR))).any(axis=1)]

# Merge the filtered numerical data with the non-numerical columns from the original DataFrame
df = pd.merge(df_no_outliers, df.select_dtypes(exclude=np.number), left_index=True, right_index=True)

df.shape
```

```python
#plot histogram for number of cars and make
plt.figure(figsize=(14,6))
df['Make'].value_counts().plot(kind='bar')
plt.title('Number of cars by make')
plt.xlabel('Make')
plt.ylabel('Number of cars')
plt.show()
```

```python
#heat maps
plt.figure(figsize=(10, 5))
c = df.corr(numeric_only=True)
sns.heatmap(c,cmap='coolwarm',annot=True)
plt.title('Correlation Heatmap')
plt.show()
```
